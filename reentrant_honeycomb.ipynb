```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from reentrant_honeycomb import *
```

```python
# rib lengths should not change while it opens
L0 = lengths(*build_lattice(theta_closed))
worst = max(np.max(np.abs(lengths(*build_lattice(t)) - L0)) for t in np.linspace(theta_closed, theta_open, 10))
print('max rib length change:', worst)
```

```python
fig, axs = plt.subplots(3, 1, figsize=(10, 10))
for ax, th in zip(axs, [theta_closed, (theta_closed + theta_open) / 2, theta_open]):
    draw(ax, th)
plt.tight_layout()
```

```python
t = np.linspace(theta_closed, theta_open, 200)
plt.plot(t, poisson(t), label='gibson-ashby')
plt.plot(t, poisson_numeric(t), '--', label='model')
plt.xlabel('theta'); plt.ylabel('poisson ratio'); plt.legend()
```
